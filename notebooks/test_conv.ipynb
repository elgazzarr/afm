```python
import flax.nnx as nnx
import jax
import jax.numpy as jnp
import jax.random as jrandom
```

```python
x = jrandom.normal(jrandom.PRNGKey(0), (7, 10, 3))
rng = nnx.Rngs(0)
conv_layer = nnx.Conv(3, 16, kernel_size=4, padding='CAUSAL', rngs=rng)
x_out = jax.vmap(conv_layer)(x)
print(x_out.shape)
```
